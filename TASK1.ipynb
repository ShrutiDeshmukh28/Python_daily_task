```python
9.
num1 = float(input("Enter first number:"))
operator = float(input("Enter operater (+ ,- ,* ,/ ,//,% ,** ):"))
num2 = float(input("Enter second number:"))

if operator == '+':
    print("Result:",num1 + num2)
elif operator == '-':
    print("Result:",num1 - num2)
elif operator == '*':
    print("Result:",num1 * num2)
elif operator == '/':
    print("Result:",num1 / num2)
elif operator == '//':
    print("Result:",num1 // num2)
elif operator == '%':
    print("Result:",num1 % num2)
elif operator == '**':
    print("Result:",num1 ** num2)
else:
    print("Invalid Operator!")
```

```python
10.
choice = input("Convert from (c)elsisius or (F)ahrenheit? Enter C or F:").upper()

if choice == 'C':
    celsius = float(input("Enter temperature in Celsius:"))
    fahrenheit = (celsis * 9/5) + 32
    print(f"{celsius}C is equal to {fahrenheit}F")
elif choice =='F':
    fahrenheit = float(input("Enter temperature in Fahrenheit:"))
    celsius = (fahrenheit - 32) * 5/9
    print(f"{fahrenheit}Fis equal to {celsius}F")
else:
    print("Invalid choice!")
```

```python
12.
print("Enter marks for 5 subject:")
m1 = float(input("subject 1:"))
m2 = float(input("subject 2:"))
m3 = float(input("subject 3:"))
m4 = float(input("subject 4:"))
m5 = float(input("subject 5:"))

total_marks = m1 + m2 + m3 + m4 + m5
percentage = (total_marks /500)* 100
print(f"percentage:{percentage}%")

if percentage >= 90:
    print("Grade: A+")
elif percentage >= 80:
    print("Grade: A")
elif percentage >= 70:
    print("Grade: B")
elif percentage >= 60:
    print("Grade: C")
else:
    print("Grade:fail")
```

```python
13.
a = input("Enter first variable(a):")
b = input("Enter second variable(b):")

# using a third variable 
temp = a
a = b
b = temp
print(f"Swapped using third variable: a = {a}, b = {b}")

# Resetting values for part b
a, b = b, a

# without third variable (python way)
a, b = b, a
print(f"Swapped the Python way: a = {a}, b = {b}")
```
