```python
12.
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(num, "x",i,"=",num * i)
    i += 1
```

```python
13.
sum = 0

num = int(input("Enter a number (0 to stop): "))

while num != 0:
    sum = sum + num
    num = int(input("Enter a number (0 to stop): "))

print("Sum of numbers =", sum)
```

```python
15.
num = float(input("Enter a positive number: "))
count = 0

while num >= 1:
    num = num / 2
    count += 1
print("Steps =", count)
```

```python
16.
num = int(input("Enter a number: "))
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

```python
18.
a = 0
b = 1
count = 1

while count <= 10:
    print(a,end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

```python
19.
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    i = 2
    prime = True

    while i < num:
        if num % i == 0:
            prime = False
            break
        i += 1

if prime:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
20.
while True:
    print("\nMenu")
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        A = int(input("Enter first number: "))
        B = int(input("Enter second number: "))
        print("Addition =", A + B)

    elif choice == 2:
        A = int(input("Enter first number: "))
        B = int(input("Enter second number: "))
        print("Subtraction =", A - B)

    elif choice == 3:
        print("Exiting program...")
        break

    else:
        print("Invalid choice")
```
