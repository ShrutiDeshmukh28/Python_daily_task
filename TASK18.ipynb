```python
34.
from abc import ABC, abstractmethod

class Vehicle(ABC):

    def __init__(self, brand, fuel_type):
        self.brand = brand
        self.fuel_type = fuel_type

    def display_vehicle(self):
        print("Brand:", self.brand)
        print("Fuel Type:", self.fuel_type)

    # Abstract Methods
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def fuel_consumption(self, distance):
        pass


class ElectricCar(Vehicle):

    def __init__(self, brand, fuel_type):
        if fuel_type != "Electric":
            raise ValueError("Fuel type must be 'Electric'.")
        super().__init__(brand, fuel_type)

    def start(self):
        print("Electric Car starts silently.")

    def fuel_consumption(self, distance):
        battery_used = distance * 0.15  
        print("Battery Consumed:", battery_used, "kWh")

    def display_vehicle(self):
        print("----- Electric Car -----")
        super().display_vehicle()


class PetrolCar(Vehicle):

    def __init__(self, brand, fuel_type):
        if fuel_type != "Petrol":
            raise ValueError("Fuel type must be 'Petrol'.")
        super().__init__(brand, fuel_type)

    def start(self):
        print("Petrol Car starts with engine ignition.")

    def fuel_consumption(self, distance):
        petrol_used = distance 
        print("Petrol Consumed:", petrol_used, "liters")

    def display_vehicle(self):
        print("----- Petrol Car -----")
        super().display_vehicle()


e = ElectricCar("Tesla", "Electric")
p = PetrolCar("Toyota", "Petrol")

e.display_vehicle()
e.start()
e.fuel_consumption(100)

print()

p.display_vehicle()
p.start()
p.fuel_consumption(100)
```

```python
5.
from abc import ABC, abstractmethod

class Report(ABC):

    def __init__(self, title, data):
        self.title = title
        self.data = data

    def header(self):
        print("Title:", self.title)
        print()

    def footer(self):
        print(" Report generation completed ")

    # Abstract Method
    @abstractmethod
    def generate(self):
        pass

        
class PDFReport(Report):

    def __init__(self, title, data):
        super().__init__(title, data)

    def generate(self):
        super().header()
        print("Generating PDF Report...")
        print("Data:", self.data)
        super().footer()


class HTMLReport(Report):

    def __init__(self, title, data):
        super().__init__(title, data)

    def generate(self):
        super().header()
        print("Generating HTML Report...")
        print("Data:", self.data)
        super().footer()


pdf = PDFReport("Sales Report", "Total Sales = $50,000")
html = HTMLReport("Student Report", "Average Marks = 85%")

pdf.generate()

print()

html.generate()
```
