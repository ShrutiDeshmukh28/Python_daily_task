### Hybrid Inheritance Questions 

```python
1.
class A:
    def method_A(self):
        print("This is Method from class A")
        
class B(A):
    def method_B(self):
        print("This is Method from class B")

class C:
    def method_C(self):
        print("This is Method from class C")

class D(B, C):
    def method_D(self):
        print("This is Method from class D")


# Create object
obj = D()
obj.method_A()
obj.method_B()
obj.method_C()
obj.method_D()
```

```python
2.
class Person:
    def set_person(self, name):
        self.name = name


class Employee(Person):
    def set_employee(self, salary):
        self.salary = salary


class Student(Person):
    def set_student(self, marks):
        self.marks = marks

class TeachingAssistant(Employee, Student):
    def set_ta(self, name, salary, marks, department):
        self.set_person(name)
        self.set_employee(salary)
        self.set_student(marks)
        self.department = department

    def display(self):
        print("\n--- Teaching Assistant Details ---")
        print("Name       :", self.name)
        print("Salary     :", self.salary)
        print("Marks      :", self.marks)
        print("Department :", self.department)


ta = TeachingAssistant()

name = input("Enter Name: ")
salary = float(input("Enter Salary: "))
marks = float(input("Enter Marks: "))
department = input("Enter Department: ")

ta.set_ta(name, salary, marks, department)
ta.display()
```

```python
3.
class Vehicle:
    def start(self):
        print("Vehicle started")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class Electric:
    def battery(self):
        print("Battery is fully charged")


class ElectricCar(Car, Electric):
    def display(self):
        print("Electric Car Details")


ec = ElectricCar()

ec.display()
ec.start()     
ec.drive()     
ec.battery()   
```

```python
4.
class A:
    def show(self):
        print("Method from class A")


class B(A):
    def show(self):
        print("Method from class B")


class C(A):
    def show(self):
        print("Method from class C")


class D(B, C):
    pass

#object
obj = D()
obj.show()
print("\nMethod Resolution Order (MRO):")
print(D.mro())
```

```python
6.
class Account:
    def __init__(self, account_number):
        self.account_number = account_number


class Savings(Account):
    def __init__(self, account_number, balance):
        super().__init__(account_number)
        self.balance = balance


class Loan(Account):
    def __init__(self, account_number, loan_amount):
        super().__init__(account_number)
        self.loan_amount = loan_amount


class SavingsLoanAccount(Savings, Loan):
    def __init__(self, account_number, balance, loan_amount):
        Savings.__init__(self, account_number, balance)
        self.loan_amount = loan_amount

    def display(self):
        print("Account Number:", self.account_number)
        print("Savings Balance:", self.balance)
        print("Loan Amount:", self.loan_amount)


# Create object
account = SavingsLoanAccount(10101, 50000, 20000)

account.display()
```

```python
7.
class Shape:
    def area(self):
        pass


class Color:
    def color(self):
        print("Color: Red")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class ColoredCircle(Circle, Color):
    def __init__(self, radius):
        super().__init__(radius)


# Create object
circle = ColoredCircle(5)

print("Area of Circle:", circle.area())
circle.color()
```

```python
10.
# Base class
class Device:
    def power_on(self):
        print("Device is powered on.")


# Derived class from Device
class Phone(Device):
    def call(self):
        print("Making a phone call.")


# Derived class from Device
class Camera(Device):
    def click(self):
        print("Taking a photo.")


# Multiple inheritance
class SmartPhone(Phone, Camera):
    def display_features(self):
        print("SmartPhone has phone and camera features.")


# Create object
mobile = SmartPhone()

mobile.power_on()
mobile.call()
mobile.click()
mobile.display_features()
```

```python
15.
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius


class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

def calculate(obj):
    print("Area =", obj.area())


# Create objects
circle = Circle(5)
rectangle = Rectangle(4, 6)

calculate(circle)
calculate(rectangle)
```

```python
19.

class A:
    def show(self):
        print("Show method from A")


class B(A):
    pass


class C(A):
    pass


class D(B, C):
    def show(self):
        print("Show method from D (Hybrid Inheritance)")


class Another:
    def show(self):
        print("Show method from Another class (Duck Typing)")


def display(obj):
    obj.show()


# Create objects
d = D()
another = Another()

display(d)
display(another)
```
