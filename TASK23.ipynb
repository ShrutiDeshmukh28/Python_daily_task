### Method Overloading Questions 

```python
1.
class Calculator:
    def add(self, *numbers):
        if len(numbers) == 2:
            return numbers[0] + numbers[1]
        elif len(numbers) == 3:
            return numbers[0] + numbers[1] + numbers[2]
        else:
            return "Enter only 2 or 3 numbers"


# Create object
calc = Calculator()

print("Sum of 2 numbers:", calc.add(10, 20))
print("Sum of 3 numbers:", calc.add(10, 20, 30))
```

```python
2.
class Area:

    def calculate(self, length, width=None):
        if width is None:
            return length * length    # Square
        else:
            return length * width     # Rectangle


# Create object
a = Area()

print("Area of Square:", a.calculate(5))
print("Area of Rectangle:", a.calculate(5, 10))
```

```python
3.
class Student:

    def details(self, name, age=None, city=None):
        if age is None:
            print("Name:", name)
        elif city is None:
            print("Name:", name)
            print("Age:", age)
        else:
            print("Name:", name)
            print("Age:", age)
            print("City:", city)


# Create object
s = Student()

s.details("Swati")
print()

s.details("Swati", 20)
print()

s.details("Swati", 20, "Hyderabad")
```

```python
4.
class MathOperations:

    def multiply(self, *args):
        result = 1

        for i in args:
            result *= i

        return result


# Create object
m = MathOperations()

print("Product of 2 numbers:", m.multiply(2, 3))
print("Product of 3 numbers:", m.multiply(2, 3, 4))
print("Product of 4 numbers:", m.multiply(2, 3, 4, 5))
print("Product of more numbers:", m.multiply(1, 2, 3, 4, 5))
```

```python
5.
class Display:

    def show(self, value):

        if isinstance(value, int):
            print("Integer:", value)

        elif isinstance(value, str):
            print("String:", value)

        elif isinstance(value, list):
            print("List:", value)

        else:
            print("Unknown data type")


# Create object
d = Display()

d.show(10)
d.show("Hello")
d.show([1, 2, 3])
```

```python
6.
class Volume:

    def calculate(self, a, b=None, c=None):

        if b is None and c is None:
            return a ** 3           # cube
        else:
            return a * b * c        # cuboid


# Create object
v = Volume()

print("Volume of Cube:", v.calculate(5))
print("Volume of Cuboid:", v.calculate(5, 4, 3))
```

```python
7.
class PrintData:

    def print_data(self, *args):
        for i in args:
            print(i)


# Create object
p = PrintData()

p.print_data("Hello")
print()

p.print_data("Python", 10, 20, "World")
```

### Method Overriding Questions 

```python
8.
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog says: Bark")


class Cat(Animal):

    def sound(self):
        print("Cat says: Meow")


d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
9.
class Vehicle:

    def start(self):
        print("Vehicle starts")


class Car(Vehicle):

    def start(self):
        print("Car starts with key")


class Bike(Vehicle):

    def start(self):
        print("Bike starts with self-start")


c = Car()
b = Bike()

c.start()
b.start()
```

```python
10.
class Employee:

    def calculate_salary(self):
        return 30000


class Manager(Employee):

    def calculate_salary(self):
        salary = 50000
        bonus = 10000
        return salary + bonus


m = Manager()

print("Manager Salary:", m.calculate_salary())
```

```python
11.
class Shape:

    def area(self):
        print("Area of Shape")


class Circle(Shape):

    def __init__(self,radius):
        self.radius = radius

    def area(self):
        print("Area of Circle:", 3.14 * self.radius * self.radius)


class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        print("Area of Rectangle:", self.length * self.width)


c = Circle(5)
r = Rectangle(4,6)

c.area()
r.area()
```

```python
12.
class BankAccount:

    def interest(self):
        print("Bank Account Interest Rate: 4%")


class SavingsAccount(BankAccount):

    def interest(self):
        print("Savings Account Interest Rate: 6%")


class CurrentAccount(BankAccount):

    def interest(self):
        print("Current Account Interest Rate: 3%")


s = SavingsAccount()
c = CurrentAccount()

s.interest()
c.interest()
```

```python
13.
class Parent:

    def display(self):
        print("This is the Parent class")


class Child(Parent):

    def display(self):
        super().display()
        print("This is the Child class")


c = Child()
c.display()
```

### Operator Overloading Question

```python
15.
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print("Point:", self.x, self.y)


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

p3.display()
```

```python
16.
class Book:

    def __init__(self, pages):
        self.pages = pages

    def __add__(self, other):
        return self.pages + other.pages

book1 = Book(100)
book2 = Book(200)

print("Total pages:", book1 + book2)
```

```python
17.
class Student:

    def __init__(self, marks):
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks

s1 = Student(85)
s2 = Student(75)

print(s1 > s2)
```

```python
18.
class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def __eq__(self, other):
        return (self.length * self.width)==(other.length * other.width)


r1 = Rectangle(10, 5)
r2 = Rectangle(5, 10)

print(r1 == r2)
```

```python
19.
class Money:

    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)
        
    def __sub__(self, other):
        return Money(self.amount - other.amount)

    def display(self):
        print("Amount:", self.amount)


m1 = Money(1000)
m2 = Money(300)

add = m1 + m2
sub = m1 - m2

add.display()
sub.display()
```

```python
20.
class Distance:

    def __init__(self, value):
        self.value = value

    def __lt__(self, other):
        return self.value < other.value

    def __gt__(self, other):
        return self.value > other.value
        
    def __eq__(self, other):
        return self.value == other.value


d1 = Distance(10)
d2 = Distance(20)

print("d1 < d2:", d1 < d2)
print("d1 > d2:", d1 > d2)
print("d1 == d2:", d1 == d2)
```
