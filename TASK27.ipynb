```python
23.
with open("notes.txt", "w") as file:
    while True:
        line = input("Enter text: ")

        if line == "STOP":
            break

        file.write(line + "\n")

print("All lines saved successfully")
```

```python
24.
with open("info.txt", "w") as file:
    file.write("Hello Python")
    
with open("info.txt", "r") as file:
    text = file.read()

encrypted = ""

for char in text:
    encrypted += chr(ord(char) + 3)

with open("encrypted.txt", "w") as file:
    file.write(encrypted)

print("File encrypted successfully")
```

```python
25.
total = 0
count = 0

with open("marks.txt", "r") as file:
    for line in file:
        mark = int(line.strip())
        total += mark
        count += 1

average = total / count

with open("average.txt", "w") as file:
    file.write("Average marks: " + str(average))

print("Average:", average)
```

```python
27.
with open("info.txt", "r") as file:
    lines = file.readlines()

with open("info.txt", "w") as file:
    for line in lines:
        if line.strip() != "":
            file.write(line)

print("Blank lines removed successfully")
```

```python
28.
with open("info.txt", "r") as file:
    lines = file.readlines()

unique_lines = []

for line in lines:
    if line not in unique_lines:
        unique_lines.append(line)

with open("unique.txt", "w") as file:
    for line in unique_lines:
        file.write(line)

print("Unique lines saved successfully")
```

```python
29.
while True:
    print("\n1. Write to file")
    print("2. Read file")
    print("3. Append to file")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        text = input("Enter text: ")

        with open("data.txt", "w") as file:
            file.write(text + "\n")

        print("Data written successfully")

    elif choice == "2":
        with open("data.txt", "r") as file:
            content = file.read()

        print("File content:")
        print(content)

    elif choice == "3":
        text = input("Enter text to append: ")

        with open("data.txt", "a") as file:
            file.write(text + "\n")

        print("Data appended successfully")

    elif choice == "4":
        print("Program exited")
        break

    else:
        print("Invalid choice")
```
