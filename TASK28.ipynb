### Level 4 : Generator with Lists

```python
16.
numbers = [10,15,20,25,30,35]
def even_numbers(nums):
    for num in nums:
        if num % 2 == 0:
          yield num

for n in even_numbers(numbers):
    print(n)   
```

```python
17.
numbers =[1,4,7,10,13,16,19]
def odd_numbers(nums):
    for num in nums:
        if num % 2 != 0:
          yield num

for n in odd_numbers(numbers):
    print(n)  
```

```python
18.
names = ["Alice","Bob","Charlie","David","Emma"]
def name_generator(names):
    for name in names:
        yield name

for name in name_generator(names):
    print(name)
```

```python
19.
numbers = [20, 55, 40, 90, 60, 15, 75]
def largest_numbers(nums):
    for num in nums:
        yield num

for n in largest_numbers(numbers):
    print(n)
```

```python
20.
words = ["Python","Java","AI","Generator","Iterator"]
def name_generator(words):
    for name in words:
        yield name

for name in name_generator(words):
    print(len(name))
```

```python
28.
def factors(num):
    for i in range(1, num + 1):
        if num % 2 == 0:
            yield i


num = 24

for factor in factors(num):
    print(factor,end = " ")
```

```python
29.
def countdown():
    for i in range(20, 0, -1):
        yield i


for num in countdown():
    print(num,end = " ")
```

```python
30.
def multiples_of_5():
    num = 5

    while True:
        yield num
        num += 5


gen = multiples_of_5()

for i in range(15):
    print(next(gen),end = " ")
```
