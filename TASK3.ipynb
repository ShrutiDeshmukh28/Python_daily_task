### find() and index() Questions

```python
14.
text = input("Enter a string: ")
ch = input("Enter a character: ")

print("First occurrence of character:", text.find(ch))
```

```python
15.
text = input("Enter a string: ")
word = input("Enter a word: ")

result = text.find(word)

print("Position:", result)
```

```python
16.
sentence = input("Enter a sentence: ")

position = sentence.index("Python")

print("Position of Python:", position)
```

```python
17.
text = input("Enter a string: ")
ch = input("Enter a character to search: ")

print("Using find():", text.find(ch))

try:
    print("Using index():", text.index(ch))
except ValueError:
    print("Using index(): Character not found (Error)")
```
