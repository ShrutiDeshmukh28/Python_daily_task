```python
30.
# 1. Load the dataset
df = pd.read_csv("C:\\Users\\lenovo\\Downloads\\Bank_customer_churn - Bank_customer_churn.csv")

# 2. Display dataset information
df.info()

# 3. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 4. Remove duplicate rows
df = df.drop_duplicates()

# 5. Rename columns
df = df.rename(columns={
    "CustomerId": "Customer_ID",
    "CreditScore": "Credit_Score",
    "EstimatedSalary": "Salary",
    "Exited": "Target"
})

# 6. Convert data types
df["Age"] = df["Age"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Target"] = df["Target"].astype(str)

# 7. Replace missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

# 8. Remove RowNumber and Surname
df = df.drop(columns=["RowNumber", "Surname"])

# 9a. Display first 10 rows
print("\nFirst 10 Rows:")
print(df.head(10))

# 9b. Display statistical summary
print("\nStatistical Summary:")
print(df.describe())

# 9c. Display value counts of Target
print("\nTarget Value Counts:")
print(df["Target"].value_counts())

# 9d. Display final dataset information
print("\nFinal Dataset Information:")
df.info()
```
