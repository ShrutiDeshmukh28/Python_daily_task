```python
10.
df["Date_of_Journey"] = pd.to_datetime(df["Date_of_Journey"],dayfirst=True)

df["Date_of_Journey"].head()
```

```python
15.
df['Dep_Time'] = pd.to_datetime(df['Dep_Time'],format='%H:%M',)
df['Dep_Hour'] = df['Dep_Time'].dt.hour
df['Dep_Hour']
```

```python
16.
df['Dep_Minute'] = df['Dep_Time'].dt.minute
df['Dep_Minute'] 
```

```python
17.
df.drop('Dep_Time', axis=1, inplace=True)
```

```python
18.
df['Arrival_Time'] = pd.to_datetime(df['Arrival_Time'],format='mixed',errors='coerce')

df['Arrival_Time'] = df['Arrival_Time'].dt.time
```

```python
19.
df["Arrival_Hour"] = pd.to_datetime(
    df["Arrival_Time"].astype(str),
    errors="coerce"
).dt.hour

df["Arrival_Minute"] = pd.to_datetime(
    df["Arrival_Time"].astype(str),
    errors="coerce"
).dt.minute

print(df[["Arrival_Hour", "Arrival_Minute"]].head())
```

```python
20.
df.drop('Arrival_Time', axis=1, inplace=True)
```

```python
21.
df['Duration_Hour'] = (
    df["Duration"]
    .str.extract(r'(\d+)\s*h',expand=False)
    .fillna(0)
    .astype(int)
)

df['Duration_Hour']
```

```python
22.
df['Duration_Minute'] = (
    df['Duration']
    .str.extract(r'(\d+)\s*m',expand=False)
    .fillna(0)
    .astype(int)
)

df['Duration_Minute']
```

```python
23.
df.drop('Duration', axis=1, inplace=True)
```

```python
24.
df['Total_Stops'] = df['Total_Stops'].replace("non-stop", 0)
df['Total_Stops']
```

```python
25.
df["Total_Stops"] = pd.to_numeric(
    df["Total_Stops"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)

df["Total_Stops"] = df["Total_Stops"].fillna(0)

df["Total_Stops"] = df["Total_Stops"].astype(int)

print(df["Total_Stops"].unique())
```

```python
26.
# Split Route using → as the separator
route_columns = df["Route"].str.split("→", expand=True)

# Create Route_1 to Route_5
for i in range(5):
    df[f"Route_{i+1}"] = route_columns[i]

print(df[["Route_1", "Route_2", "Route_3", "Route_4", "Route_5"]].head())
```
