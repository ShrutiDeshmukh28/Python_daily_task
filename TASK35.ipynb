```python
5.
df['missing_values']=df.isnull().sum()/len(df) * 100
df['missing_values']
```

```python
6.
missing_id = df['Customer ID'].isna().sum()
print(missing_id)
```

```python
7.
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print(df.columns)
```

```python
8.
df.duplicated()
```

```python
9.
df.drop_duplicates()
```

```python
10.
df['customer_segment'].unique()
```

```python
11.
df['customer_segment'] = (
    df['customer_segment']
    .astype(str)
    .str.strip()
    .str.lower()
    .str.title()
)

print("\nUnique values after cleaning:")
print(df['customer_segment'].unique())
```

```python
12.
print(df['gender'].unique())

df['gender'] = df['gender'].astype(str).str.strip().str.lower()

df['gender'] = df['gender'].replace({
        'm': 'Male',
        'male': 'Male',
        'f': 'Female',
        'female': 'Female'
    })

print(df['gender'].unique())
```

```python
13.
print(df['is_active'].unique())

df['is_active'] = (
    df["is_active"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        'yes': 1,
        'y': 1,
        '1': 1,
        'no': 0,
        'n': 0,
        '0': 0
    })
)

print(df['is_active'].unique())
```

```python
14.
df['churn_flag'] = df['churn_flag'].map({'Yes' : 1, 'No' : 0})
df.head(10)
```

```python
15.
invalid_income = df[
    pd.to_numeric(df["income_usd"],errors="coerce").isna()
    & df["income_usd"].notna()
]["income_usd"]
print(invalid_income.unique())
```
