```python
13.
titanic["age_group"] = pd.cut(
    titanic["age"],
    bins=[0, 18, 60, float("inf")],
    labels=["Child", "Adult", "Senior"],
    right=False
)

sns.catplot(
    data=titanic,
    x="age_group",
    hue="survived",
    kind="count"
)
plt.xlabel("Age Group")
plt.ylabel("Passenger Count")
plt.show()
```

```python
14.
sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    hue="sex",
    col="survived",
    kind="bar"
)
plt.show()
```

```python
15.
g= sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    height=5,
    aspect=1.1,
    palette="Set2"
)

g.set_axis_labels("Passenger Class", "Fare")
g.set_titles("Survived = {col_name}")
g.fig.suptitle("Fare Distribution by Class, Gender and Survival", y=1.05)

plt.show()
```

```python
30.
sns.set_theme(style="whitegrid")

g = sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    kind="box",
    palette="Set2",
    height=5,
    aspect=1.1
)

g.set_axis_labels("Day", "Total Bill ($)")

g.set_titles("{col_name}")

g.figure.suptitle(
    "Total Bill by Day, Sex, and Time",
    fontsize=16,
    fontweight="bold",
    y=1.03
)

g._legend.set_title("Sex")

plt.show()
```
