### Categorical Columns (Inconsistencies)

```python
13.
department = (
    df["Department"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+","",regex = True)
    .str.casefold()
)

corrections = {
    "sale": "Sales",
    "financ": "Finance",
    "hr": "HR",
    "hr": "HR",
    "it": "IT"
}
df["Department"] = department.replace(corrections)

df["Department"] = df["Department"].str.title()

df["Department"] = df["Department"].replace({

    "Hr" : "HR",
    "It" : "IT",
})

unique_department_count = df["Department"].nunique(dropna = True)

print("Unique departments after cleaning :", unique_department_count)
print(df["Department"].value_counts())
```

```python
14.
city = (
    df["City"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+","",regex = True)
    .str.casefold()
)

corrections = {
    "Los Angles": "Los Angeles",
    "Losangles": "Los Angeles",
    "Ny": "New York",
    "Philly": "Philadelphia",
    "Sandiego": "San Diego",
}
df["City"] = city.replace(corrections)
df["City"] = df["City"].str.title()
df["City"] = df["City"].replace({

    "Losangles": "Los Angeles",
    "Newyork": "New York",
    "Phoenix": "Philadelphia",
    "Sanjose": "San Diego",
    "Houstontx": "Houston"
})

unique_city_count = df["City"].nunique(dropna = True)

print("Unique city after cleaning:",unique_city_count)
print(df["City"].value_counts())
```

```python
15.
original_gender = df["Gender"].copy()

gender_normalized = (
     original_gender.astype("string")
    .str.strip()
    .str.casefold()
)

gender_mapping = {
    "M": "Male",
    "Male": "Male",
    "F": "Female",
    "Female": "Female",
    "O": "Other",
    "Other": "Other"
}

cleaned_gender = gender_normalized.map(gender_mapping)

# convert non-missing, unrecognized values to other
cleaned_gender = cleaned_gender.where(
    original_gender.isna(),
    cleaned_gender.fillna("other")
)

# count records whose value changed 
changed_count = (
    original_gender.notna()
    &(cleaned_gender.astype("string") != original_gender.astype("string"))
).fillna(False).sum()

df["Gender"] = cleaned_gender

print("Records changed:",changed_count)
print("Final values:",df["Gender"].dropna().unique())
```

### Performance_Score & Is_Active

```python
16.
df["Performance_Score"] = pd.to_numeric(
    df["Performance_Score"],
    errors="coerce"
)

mean_score = df["Performance_Score"].mean()

df["Performance_Score"] = df["Performance_Score"].fillna(mean_score)

print("Mean performance score:", mean_score)
print("Remaining missing values:", df["Performance_Score"].isna().sum())
```

```python
17.
active_mapping = {
    "yes": True,
    "y": True,
    "1": True,
    "active": True,
    "true": True,
    
    "no": False,
    "n": False,
    "0": False,
    "inactive": False,
    "false": False
}

normalized_active = (
    df["Is_Active"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

df["Is_Active"] = normalized_active.map(active_mapping).astype("boolean")

print(df["Is_Active"].value_counts(dropna=False))
```

### Email & Name

```python
18.
email = df["Email"].astype("string").str.strip()

email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

valid_email = email.str.match(email_pattern,na=False)

df["Invalid_Email"] = ~valid_email

invalid_count = df["Invalid_Email"].sum()

print("Invalid email addresses:", invalid_count)
```

```python
19.
name = df["Name"].astype("string").str.strip()

missing_names = (
    name.isna()
    | name.str.casefold().isin(["", "N/A", "Unknown"])
)

df["Name"] = name.mask(missing_names,pd.NA)

missing_name_count = df["Name"].isna().sum()

print("Names still missing:", missing_name_count)
```

### Final Challenge

```python
20.
# 1. Clean Age 
df["Age"] = pd.to_numeric(df["Age"],errors="coerce")
df["Age"] = df["Age"].fillna(df["Age"].median())

# Clean Salary
df["Salary"]=(
    df["Salary"]
    .astype("string")
    .str.replace(r"[$]","",regex = True)
    .str.strip()
)
df["Salary"] = pd.to_numeric(df["Salary"],errors="coerce")

# 3.Cleam Join_Data and extract year
df["Join_Data"] = pd.to_datetime(
    df["join_Data"],
    errors="coerce",
    format = "mixed"
)
df["Join_Year"] = df["Join_Data "].dt.year

# 4.Clean Department
department = (
    df["Department"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+","",regex = True)
    .str.casefold()
)

department_map = {
    "sale": "Sales",
    "sals": "Sales",
    "financ": "Finance",
    "financ": "Finance",
    "hr": "HR",
    "it": "IT"
}

df["Department"] = department.replace(department_map).str.title()

df["Department"] = df["Department"].replace({

    "Hr" : "HR",
     "It": "IT"
})

# 5. Clean City
city = (
    df["City"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+","",regex = True)
    .str.casefold()
)

city_map = {
    "los angles": "Los Angeles",
    "los angles": "Los Angeles",
    "la": "Los Angeles",
    "ny": "New York",
    "new york city": "New York",
    "Philly": "Philadelphia",
    "Philadelphia": "Philadelphia"
} 
df["City"] = city.replace(city_map).str.title()

# 6. Clean Gender
gender = (
    df["Gender"]
    .astype("string")
    .str.strip()
    .str.casefold()
)
gender_map = {
    "male": "Male",
    "m": "Male",
    "man": "Male",
    "female": "Female",
    "f": "Female",
    "women": "Female",
    "other": "Other",
    "o": "Other",
    "non-binary": "Other",
    "nonbinary": "Other",
    "nb": "Other"
}

df["Gender"]=gender.map(gender_map)
df["Gender"]=df["Gender"].fillna(
    gender.where(gender.isna(),"Other")
)

# 7. Clean Performance_Score
df["Performance_Score"] = pd.to_numeric(
    df["Performance_Score"],
    errors="coerce"
)

df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)

# 8. Clean Is_Active
active = (
    df["Is_Active"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

active_map = {
    "yes": True,
    "y": True,
    "1": True,
    "active": True,
    "true": True,
    
    "no": False,
    "n": False,
    "0": False,
    "inactive": False,
    "false": False
}
df["Is_Active"] = active.map(active_map).astype("boolean")

# 9. Flag invalid email addresses
email = df["Email"].astype("string").str.strip()
email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

df["Invalid_Email"] = (
    email.notna()
    &~email.str.match(email_pattern,na=False)
)

# 10. Clean Name
name = df["Name"].astype("string").str.strip()

invalid_names = (
    name.isna()
    | name.str.casefold().isna(["", "n/a", "unknown"])
)

df["Name"] = name.mask(invalid_names,pd.NA)

# Save the cleaned dataset 
df.to_csv("cleaned_dataset.csv",index=False)

# Report results 
print("Final shape:",df.shape)
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
print("Saved as: cleaned_dataset.csv")
```
