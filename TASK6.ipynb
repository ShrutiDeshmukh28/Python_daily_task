```python
23.
student_records = {}

for i in range(4):
    print(f"---Student {i + 1}---")
    name = input("Enter student name:")
    marks = float(input(f"Enter marks for {name}:"))
student_records [name] = marks 
highest_student = max(student_records,key=student_records.get)
highest_marks = student_records[highest_student]

print("\n---Student Records---")
print(student_records)
print(f"\nTop Scorer: {highest_student} with {highest_marks} marks")
```
