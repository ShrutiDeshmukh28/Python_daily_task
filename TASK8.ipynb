```python
15.
a = float(input("Enter length of side 1 :"))
b = float(input("Enter length of side 2 :"))
c = float(input("Enter length of side 3 :"))

if (a + b > c) and (a + c > b) and (b + c > a):
          print("The given sides form a valid triangle")

          # step 2: Classify the Triangle
          if a == b == c:
              print("Type: Equilateral Triangle")
          elif a == b or b == c or a == c:
              print("Type: Isosceles Triangle")
          else:
              print("Type: Scalene Triangle")
else: 
    print("The given sides DO NOT form a valid triangle ")
```
