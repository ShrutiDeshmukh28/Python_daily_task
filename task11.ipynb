```python
11.
n = int(input("Enter number:"))
rev = 0
while n>0:
   digit = n % 10
   rev = rev *10 + digit
   n = n//10
print("Reverse=",rev)
```

```python
33.
for num in range(10, 51):
    if num == 50:
        break

    if num < 2:
        continue

    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num)
```
