### Args (Multiple Positional Arguments)

```python
19.
def add_all(*number):
    c=sum(number)
    return c
add_all(2+3+5+3+6+9+8+7+4+5+1+0)
```

```python
20.
def print_all(*args):
    return args
print_all("shruti","disha",18,19,True)
```

### kwargs (Multiple Keyword Arguments)

```python
26.
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(Name="Rahul", Age=20, Course="Python", Marks=85)
```

```python
27.
def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

create_profile(Name="Rahul", Age=21, City="Hyderabad", Course="Python", Grade="A")
```

```python
28.
def calculate_total(**kwargs):
    total = 0

    for item, price in kwargs.items():
        print(item, ":", price)
        total = total + price

    print("Total =", total)

calculate_total(Bread=40, Milk=30, Rice=100, Fruits=80)
```

```python
29.
def display_info(name, **kwargs):
    print("Name:", name)

    for key, value in kwargs.items():
        print(key, ":", value)

display_info("Shruti", Age=22, City="Hyd", Course="Data Science", Marks=90)
```

```python
30.
def save_data(**kwargs):
    print("Saving data:")

    for key, value in kwargs.items():
        print(key, ":", value)

save_data(Name="Shruti", Age=23, City="Hyderabad", Course="Data Science")
```
