```python
9.
def detect_copy_type(copy_function): 
    "Detects whether a copy function perform a deep copy,shallow copy or assignment "
    original = [[0]]
    copied = copy_function(original)
    copied[0][0]=99

    is_same_top_level = original is copied
    is_nested_modified = original[0][0] == 99

    if is_same_top_level:
        return"assignment(no copy)"
    elif is_nested_modified:
        return"shallow copy"
    else:
        return"deep copy"

if __name__ == "__main__":
    import copy

    assign_fn = lambda x:x
    print(f"Assignment:{detect_copy_type(assign_fn)}")

    shallow_fn = lambda x:x.copy()
    print(f"Shallow Copy:{detect_copy_type(shallow_fn)}")

    deep_fn = lambda x: copy.deepcopy(x)
    print(f"Deep Copy:{detect_copy_type(deep_fn)}")
```

```python
3.
class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def apply_discount(self):
        discount_amount = self.price * 0.10
        self.price -= discount_amount
        return self.price
        
my_phone = Mobile("Samsung", "Galaxy S23", 50000)

print(f"Original Price:{my_phone.price}")
my_phone.apply_discount()
print(f"Price after 10% discount: {my_phone.price}")
```

```python
5.
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited", amount)
        print("Balance is", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn", amount)
            print("Balance is", self.balance)
        else:
            print("Insufficient balance")


account1 = BankAccount("Rahul", 5000)

account1.deposit(2000)
account1.withdraw(3000)
account1.withdraw(5000)
```

```python
7.
class Temperature:
    def __init__(self, temperature):
        self.temperature = temperature

    def set_temperature(self, value):
        if -20 <= value <= 50:
            self.temperature = value
            print("Temperature:", self.temperature)
        else:
            print("Invalid")
            
temp = Temperature(25)
temp.set_temperature(40)
temp.set_temperature(60)
```

```python
8.
class MovieTicket:
    def __init__(self, movie_name, seats):
        self.movie_name = movie_name
        self.seats = seats

    def book_ticket(self, number):
        if number <= self.seats:
            self.seats -= number
            print("Booked Successfully")
            print("Remaining Seats:",self.seats)
        else:
            print("Seats not available")

ticket = MovieTicket("Pushpa 2", 50)
ticket.book_ticket(5)
```

```python
10.
class Product:
    def __init__(self, product_name, quantity):
        self.product_name = product_name
        self.quantity = quantity

    def sell(self, qty):
        if qty <= self.quantity:
            self.quantity -= qty
            print("Remaining Quantity:",self.quantity)
        else:
            print("Not enough Stock")

product = Product("Laptop",20)
product.sell(3)
```

```python
19.
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
            self.balance += amount
            print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid Amount")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print("Balance:",self.balance)
            
atm = ATM(5000)
atm.withdraw(0)
atm.withdraw(12000)
atm.withdraw(2000)
```

```python
21.
class InventorySystem:
    def __init__(self, product_name, stock):
        self.product_name = product_name
        self.stock = stock

    def update_stock(self, change):
            if self.stock + change >= 0:
                self.stock += change
                print("Stock:", self.stock)
            else:
                print("Stock cannot become negative")
                
inventory = InventorySystem("Laptop", 50)
inventory.update_stock(-10)
```

```python
22.
class EmployeePerformance:
    def __init__(self, employee_name, rating):
        self.employee_name = employee_name
        self.rating = rating

    def update_rating(self, new_rating):
        if 1 <= new_rating <= 5:
            self.rating = new_rating
            if self.rating == 5:
                print("Excellent")
            else:
                print("Rating:", self.rating)
        else:
            print("Invalid Rating")

employee = EmployeePerformance("Sumit", 4)
employee.update_rating(5)
```

```python
23.
class StudentAttendance:
    def __init__(self, total_classes, attended_classes):
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    def mark_attendance(self, number):
        if self.attended_classes + number <= self.total_classes:
            self.attended_classes += number
            percentage = (self.attended_classes / self.total_classes) * 100
            print("Attendance:", percentage, "%")
        else:
            print("Invalid Attendance")
            
attendance = StudentAttendance(100,60)
attendance.mark_attendance(20)
```

```python
24.
class Bus:
    def __init__(self, bus_name, total_seats, booked_seats):
        self.bus_name = bus_name
        self.total_seats = total_seats
        self.booked_seats = booked_seats

    def reserve(self, number):
        if self.booked_seats + number <= self.total_seats:
            self.booked_seats += number
            print("Remaining Seats:", self.total_seats - self.booked_seats )
        else:
            print("Reservation Not Possible")
            
bus = Bus("Express", 50,20)
bus.reserve(10)
```
