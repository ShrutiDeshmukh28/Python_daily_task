### String and String Slicing Questions

```python
5.
Name = input ("Enter your Full Name:")

print("1. Uppercase:", Name.upper())
print("2. Lowercase:", Name.lower())
print("3. Length of Name:", len(Name))
print("4. First 3 Characters:", Name[:3])
print("5. Last 3 Characters:", Name[-3:])
```

```python
6.
sentence = input("Enter any sentence:")
print("First 10 characters:", sentence[:10])
print("Last 10 characters:", sentence[:-10])
print("Characters from index 5 to 15:", sentence[5:16])
print("Every 2nd characters:", sentence[::2])
```

```python
7.
word = input ("Enter a word:").lower()

if word == word [::-1]:
    print("The word is palindrome")
else:
    print("The word is not a palindrome")
```

```python
12.
city = input("Enter your city name:")

#1. Reverse order 
print("Reverse order :",city[::-1])

#2. Count vowels
vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in city if char in vowels)
print("2. Number of vowels:",vowel_count)

#3. without first and last character
print("3.Trimmed city name:",city[1:-1])
```
